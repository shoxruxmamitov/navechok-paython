```python
import numpy as np
```

massiv yaratib olamiz

```python
ilik = np.arange(31) # 0 va 30 gacha bolgan massiv yaratamiz
```

```python
np.save('latta', ilik) # bir bir ozgaruvchi (latta) yaratib olamiz
```

```python
ilik1 = np.load('latta.npy') # yana ozgaruvchi (ilik1) yaratib olamiz
```

```python
lik = np.arange(20) # massiv yarartib olamiz
```

```python
np.savez('orerblok', t = ilik , k = lik) # orerblokka (t) va (k) ozgaruchiga sax qilib olamiz
```

```python
orerblok = np.load('orerblok.npz')
```
