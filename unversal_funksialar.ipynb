modf # bu funksia sonlarni qoldiq va butun ni ajratib beradi

```python
tasodif_son=np.random.randn(6)
tasodif_son
```

```python
qoldiq, butun = np.modf(tasodif_son)
```

```python
print(qoldiq) # qoldiq sonni ajratib berdi
print(butun)  # butun sonni ajratib berdi
```
